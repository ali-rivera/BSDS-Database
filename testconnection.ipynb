```python
from datasets.models import Dataset
import pymongo
import datetime
import pytz #for EST

myclient = pymongo.MongoClient(f"mongodb://{MONGO_INITDB_ROOT_USERNAME}:{MONGO_INITDB_ROOT_PASSWORD}@localhost:27017/")
```

```python
#create an entry
datasets = [
Dataset(name="states_all_2_training.csv", description = "Inserting first test data set", file_url="https://github.com/UVADS/DS-3021/blob/main/Kaggle/states_all_2_training.csv", uploaded_at = datetime.datetime.now()),
Dataset(name="bank.csv", description = "Inserting second test data set", file_url="https://github.com/UVADS/DS-3021/blob/main/07_ML_Eval_Metrics/bank.csv", uploaded_at = datetime.datetime.now()),   
]
for dataset in datasets:
    # dataset.uploaded_at = datetime.datetime.now(eastern)  # Ensure EST timezone
    dataset.save()
```

```python
#fetch and print datasets
eastern = pytz.timezone("America/New_York") #EST not working

for dataset in Dataset.objects:
    est_time = dataset.uploaded_at.replace(tzinfo=pytz.utc).astimezone(eastern)
    print(f"Name: {dataset.name}, Uploaded At (EST): {est_time.strftime('%Y-%m-%d %I:%M %p %Z')}")
```

```python
#delete
dataset = Dataset.objects(name="states_all_2_training.csv").first()
if dataset:
    dataset.delete()
    print("Dataset deleted.")
else:
    print("Dataset not found.")
```
